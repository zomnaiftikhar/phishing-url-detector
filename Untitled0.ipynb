```python
import pandas as pd, numpy as np, json, os
import tensorflow as tf
print(tf.__version__)  # note this for requirements.txt

df = pd.read_csv("phishing_site_urls.csv")
df.columns = ["url", "label"]
df = df.drop_duplicates(subset="url").dropna()   # dedupe BEFORE splitting to avoid leakage
df["y"] = (df["label"] == "bad").astype(int)
print(df.shape, df["y"].value_counts())
```

```python
from sklearn.model_selection import train_test_split

train, temp = train_test_split(df, test_size=0.30, stratify=df["y"], random_state=42)
val, test = train_test_split(temp, test_size=0.50, stratify=temp["y"], random_state=42)

MAXLEN = 200
chars = [chr(i) for i in range(32, 127)]            # printable ASCII
char2idx = {c: i + 1 for i, c in enumerate(chars)}   # 0 = padding, 96 = unknown

def encode(urls):
    out = np.zeros((len(urls), MAXLEN), dtype="int32")
    for i, u in enumerate(urls):
        for j, c in enumerate(str(u)[:MAXLEN]):
            out[i, j] = char2idx.get(c, len(chars) + 1)
    return out

Xtr, ytr = encode(train["url"]), train["y"].values
Xva, yva = encode(val["url"]), val["y"].values
Xte, yte = encode(test["url"]), test["y"].values
```

```python
from tensorflow.keras import layers, models

model = models.Sequential([
    layers.Input(shape=(MAXLEN,)),
    layers.Embedding(len(chars) + 2, 32),
    layers.Conv1D(128, 5, activation="relu"),
    layers.GlobalMaxPooling1D(),
    layers.Dense(64, activation="relu"),
    layers.Dropout(0.3),
    layers.Dense(1, activation="sigmoid"),
])
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

neg, pos = np.bincount(ytr)
cw = {0: len(ytr) / (2 * neg), 1: len(ytr) / (2 * pos)}   # handles class imbalance

history = model.fit(Xtr, ytr, validation_data=(Xva, yva),
                    epochs=2, batch_size=256, class_weight=cw)
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt

pred = (model.predict(Xte) > 0.5).astype(int).ravel()
print(classification_report(yte, pred, target_names=["good", "phishing"]))
print(confusion_matrix(yte, pred))

plt.plot(history.history["accuracy"], label="train")
plt.plot(history.history["val_accuracy"], label="val")
plt.legend(); plt.title("Accuracy"); plt.savefig("training_curve.png")
```

```python
os.makedirs("model", exist_ok=True)
model.save("model/phishing_model.keras")
json.dump({"maxlen": MAXLEN, "char2idx": char2idx, "unk": len(chars) + 1},
          open("model/vocab.json", "w"))
```
